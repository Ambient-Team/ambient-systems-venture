#### Data Preprocessing
- Group entries into regular time intervals (e.g., consolidate readings into every 5 minutes)
- Join data-sets (on same time interval): (x, y)
    where x is the (vector, multi-dimensional) features (e.g., input 1, input 2)
    where y is the (scalar, 1 dimensiona) output (e.g., sensor reading or plant characteristic)
- Create design matrix (where first column is for the hyper-parameter)
- Standardization and normalization

#### Training Functions
- Loss function: calculates error between true y-values and mean squared error (MSE)
- Gradient descent: fits the linear regression model using Gradient Descent

#### Model selection
- Apply different types of models to find the ones that fit the best for the given data
- Apply multivariate regression algorithms: LinearRegression (and related), KNeighborsRegressor, DecisionTreeRegressor, RandomForestRegressor (and related)

#### Training (Hyper-parameter Optimization)
- Running the gridsearch function to try different combinations of hyper-parameters
- Calculates accuracy and scores

#### Presentation
- Output: model function, hyper-parameters, recommended inputs given desired outputs
